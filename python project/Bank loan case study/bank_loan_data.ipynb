```python
def uni_numerical_analysis(dataframe,column):
    sns.set(style="darkgrid")
    plt.figure(figsize=(25,10))
    plt.subplot(1,3,1)
    sns.boxplot(data=dataframe,x=column,orient="v")
    plt.title(column)
    plt.subplot(1,3,2)
    sns.histplot(dataframe[column],color="purple",kde=True)
    plt.title(column)
    plt.show()
```

```python
for i in numerical_cols:
    uni_numerical_analysis(df_app,i)
```

```python
def uni_categorical_columns(dataframe,column):
    sns.set(style="darkgrid")
    plt.figure(figsize=(15,8))
    sns.barplot(x=dataframe[column].value_counts(),y=dataframe[column].value_counts().index,palette="magma")
    plt.xlabel("frequency")
    plt.ylabel(column)
    plt.show()
    
```

```python
for i in categorical_cols:
    uni_categorical_columns(df_app,i)
```
