```python
#LIST
l=[]
print(l)
print(type(l))
```

```python
l=[10,20,30,"hello",10.5]
print(l)
print(type(l))
print(l[0])   #indexed
l[0]=1000     #mutable
print(l)
```

```python
#indexing and slicing
l=[10,20,30,"hello",10.5,1000,200,300,100]
print(l[3])
print(l[3][1])
print(l[-3])
print(l[::-1])
print(l[::])
print(l[:6:2])
print(l[-1:-7:-2])
print(l[-1:7:-2])
```

```python
#direct iteration
l=[10,20,30,"hello",1000,10.5,11] 
for i in l:
    print(i)
```

```python
#indirect iteration
l=[10,20,30,"hello",1000,10.5,11] 
for i in range(len(l)):
    print(l[i])
```

```python
#function
# len()
# max()
# min()
l=[10,20,30,"hello",1000,10.5,11] 
print(len(l))
count=0
for i in l:
    count= count+1
print(count)    
```

```python
#max()
# min()
l=[10,20,30,1000,10.5,11]
print(max(l))
print(min(l))
```

```python
l=[10,20,30,1000,10.5,11]
max_value=l[0]
for i in l:
    if i>max_value:
        max_value=i
print(max_value)        
```

```python
l=[10,20,30,1000,10.5,11]
min_value=l[0]
for i in l:
    if i<min_value:
        min_value=i
print(min_value)   
```

```python
l=[10,20,30,1000,10.5,11]
print(sum(l))
sum=0
for i in l:
    sum=sum+i
print(sum)    
```

# METHODS

```python
#ADD ELEMENT TO AN EXISTING LIST.
# APPEND
# listobj.append(value/obj)
# append method- adds new element at last position.  that is at the index value of -1
l=[10,20,30,"hello",10.5]
l.append(1000)
print(l)
```

```python
l=[10,20,30,"hello",10.5]
l.append([100,200,300])
print(l)
```

```python
l=[]
for i in range(5):
    n=int(input("enter any number"))
    l.append(n)
    print(l)
print(l)    
    
```

```python
l=[100,23,34,11,17,32,35,98,12,7]
l1=[]
l2=[]
for i in l:
    if i%2==0:     #condition for even numbers
        l1.append(i)
    else:
        l2.append(i)
print(l1)    
print(l2)
```

```python
l=[]
for i in range(0,8,2):
    l.append(chr(65+i))
print(l)    
```

```python
#insert
#listobj.insert(position,value)
l=[10,20,30,"hello",10.5]
l.insert(2,"decode")
print(l)
```

```python
#extend
#listobj.extend(iterable)
l=[10,20,"decode",30,"hello"]
l.extend([100])
print(l)
l=[10,20,"decode",30,"hello"]
l.extend("decode")
print(l)
```

```python
l=["naman","decode","nitin","bhopal","data","madam"]
l1=[]
for i in l:
    p=i
    for j in i:
        rev=i[::-1]
    if p==rev:
            l1.append(i)
print(l1)
            
```

```python
l=["naman","decode","nitin","bhopal","data","madam"]
l1=[]
for i in l:
    p=i
    rev=i[::-1]
    if p==rev:
            l1.append(i)
print(l1)
```

```python
#index
#listobj.index(value,[start],[stop])
#bydefault - 0 stop by default - end of the list
l=[10,20,30,"hello",10,10.5,1000,10,30,10,100]
print(l.index(10))
print(l.index(10,2))
print(l.index(10,1,4)) # will throw error cause it will run till n-1=3
```

```python
#count
#listobj.count(element)
l=[10,20,30,"hello",10,10.5,1000,10,30,10,100]
print(l.count(100))

```

```python
#reverse()
#listobj.reverse()
l=[100,10,20,9]
l.reverse()
print(l)
```

```python
l=[10,20,30,40]
for i in range(len(l)):
    for j in range(i+1,len(l)):
        l[i],l[j]=l[j],l[i]
print(l)
    
```
