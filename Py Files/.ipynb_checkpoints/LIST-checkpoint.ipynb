```python
#LIST-
l=[]
print(l)
print(type(l))
```

```python
l=[10,20,30,"hello",10.5]
print(l)
print(type(l))
print(l[0]) # indexed
l[0]=1000
print(l)
```

```python
#indexing and slicing
l=[1,2,3,"hello",10.5,2.,4,4,]
print(l)
print(l[3])
print(l[-3])
print(l[: :-1]) # by default -1 stop - end of string
```

```python
l=[10,20,30,"hello",1000,10.5,11]
for i in l:
    print(i)
```

```python
# method
# add element to an existing list
# append
# listobj.append(value/obj)
# append method add new element at last position.. -1.
l=[10,20,30,"hello",10.5]
l.append(1000)
print(l)
```

```python
l=[10,20,30,"hello",10.5]
l.append([100,200,300])
print(l)
```

```python
l=[]
for i in range(5):
    n=int(input("enter any number: "))
    l.append(n)
print(l)
```

```python
# [4,16,36,64,100]
l=[]
for i in range(2,11,2):
    l.append(i**2)
print(l)
```

```python
l=[5,3,2,6,1,4]
l1=[]
for i in l:
    l1.append(i**3)
print(l1)
```

```python
l=[100,23,34,11,17,32,35,98,12,7]
# output - l1=[100,34,32,98,12]
l1=[]
l2=[]
for i in l:
    if i%2==0:
        l1.append(i)
    else:
        l2.append(i)
print(l1)
print(l2)
```

```python
l=[]
for i in range(0,8,2):
    l.append(chr(65+i))
print(l)
```

```python
l=[10,20,30,40,[100,200,300,"decode"],1000,2000]
l1=["delhi","mumbai","indore"]
# output-
# [10,20,30,40,[100,200,300,"decode","delhi","mumbai"],1000,2000] 
for i in l1:
    l[4].append(i)
print(l)
```
