```python
l=[1,5,3,4,5,1,6,7,7,5,9,11,5,8]
#median
sort_l=sorted(l)
print(sort_l)
n=len(l)
if n%2!=0:
    median_odd=sort_l[n//2]
else:
    median_even=sort_l[n//2]+sort_l[(n//2) - 1]/2
print(median_odd," median in odd elements in a list")
print(int(median_even)," median in odd elements in a list")
```
