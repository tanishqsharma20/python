```python
l1=[5,1,3,4,2]
l2=[2,5,1,3,4]
l3=[]
for i in range(len(l1)):
    l3.append(l1[i]**l2[i])
print(l3)
    
```

```python
l=["madam","aa","decode","comaas","naman","figure"]
l1=[]
for i in l:
    main=i
    rev=""
    for j in i:
        rev=j+rev
    if rev==main:
            l1.append(main)
print(l1)
    
```
